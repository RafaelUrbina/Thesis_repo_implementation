## Satellite query

```sql
TITLE-ABS-KEY ( 
  ( satellite* OR drone* OR uav OR "remote sensing" OR "aerial imagery" OR "SAR" )
  AND 
  ( "water pipeline" OR "water network" OR "water main" OR "buried pipe" OR "aqueduct" )
  AND 
  ( detect* OR map* OR trac* OR rout* OR alignment OR location ) 
  ) 
  AND NOT 
  ( "oil" OR "gas" OR "data pipeline" OR "marine" OR "ocean" )
```

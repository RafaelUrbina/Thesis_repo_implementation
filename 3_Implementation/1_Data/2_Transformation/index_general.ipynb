## Reference table describing what each layer represents conceptually. the layers are grouped by their thematic domain.

### **1. Base Framework & Administrative Boundaries**

*These layers define geographical study area constraints and the target grid structure.*

| Layer Name | Geometry | Conceptual Description |
| --- | --- | --- |
| **`h310grid`** | Polygon | The master resolution-10 H3 hexagonal spatial index grid used as the uniform data container for the entire analysis. |
| **`provcm01012026_wgs84`** | Polygon | The official geographic boundary of the Metropolitan City (Province) of Florence, defining the spatial limit of the study. |
| **`com01012026_wgs84`** | Polygon | Municipal (communal) borders within Florence, useful for local administrative grouping and comparing policy differences. |

### **4. Geotechnical Dynamics & Landslide Hazards**

*Layers documenting ground stability, historical landslide inventories, and active soil movement vectors.*

| Layer Name | Geometry | Conceptual Description |
| --- | --- | --- |
| **`frane_piff_toscana_opendata`** | Point | Historical landslide event inventory marking individual landslide failure points and their exact movement classification types. |
| **`frane_poly_toscana_opendata`** | Polygon | Mass-movement inventory mapping the physical spatial boundaries and footprints of documented landslide features. |
| **`franosita__di_superficie_interessata_da_frane`** | Polygon | Regional classification zones defining land susceptibility based on the historical percentage of the area disrupted by mass movements. |
| **`velocita_movimento_nel_piano`** | Polygon | Monitored deformation zones outlining the velocity and orientation of slow ground displacement. |
| **`celle_soli_PS_descendenti`** | Polygon | Specifically mapped spatial cells showing downslope or downward vertical ground deformation. |
| **`celle_soli_PS_ascendenti`** | Polygon | Specifically mapped spatial cells showing upward vertical ground deformation or soil swelling. |
| **`seismic_points_utm32n`** | Point | Historical database of seismic epicenters and earthquake events, recording calculated moment magnitudes (Mw) and associated measurement errors. |
| **`mosaicatura_ispra_2024_pericolosita_frana_pai`** | Polygon | **National Landslide Hazard Mosaic (PAI):** Official hydrogeological risk framing classifying active and potential landslide surfaces into progressive structural classes spanning from Attention Zones (`AA`) up through low-to-extreme hazard states (`P1` to `P4`). |

### **5. Pedological (Soil) Properties & Agricultural Capacity**

*Physical and chemical properties of the soil profile that govern water infiltration, erosion, and structural stability.*

| Layer Name | Geometry | Conceptual Description |
| --- | --- | --- |
| **`gruppo_idrologico_usda`** | Polygon | USDA Hydrologic Soil Groups (A, B, C, D) classifying soil types by their minimum infiltration rate and runoff potential when thoroughly wet. |
| **`drenaggio_interno`** | Polygon | The natural internal drainage capability of the soil matrix, tracking how efficiently water moves downward through the profile. |
| **`ksat_30__conducibilita_idraulica_satura_sezione_030_cm`** | Polygon | Saturated hydraulic conductivity ($K_{sat}$) in the shallow topsoil horizon (0–30 cm), defining how fast water flows through saturated surface layers. |
| **`ksat_150__conducibilita_idraulica_satura_sezione_0150_cm`** | Polygon | Saturated hydraulic conductivity ($K_{sat}$) deep within the subsoil profile (0–150 cm), indicating deeper drainage properties. |
| **`awc__available_water_capacity`** | Polygon | Available Water Capacity ($AWC$), measuring the maximum volume of moisture a soil profile can retain for plant roots before draining away. |
| **`profondita_utile_per_le_radici_cm`** | Polygon | Effective soil depth unobstructed by bedrock or hardpans, limiting or allowing root penetration. |
| **`pietrosita_superficiale_`** | Polygon | Surface stoniness coverage class, tracking the presence of loose, detached rock fragments on top of the soil horizon. |
| **`rocciosita_`** | Polygon | Surface bedrock outcroppings, mapping zones where solid, continuous bedrock is completely exposed without soil cover. |
| **`fertilita_chimica_dellorizzonte_superficiale`** | Polygon | Baseline chemical fertility status of the surface horizon, tracking the natural nutrient capacity of the topsoil. |
| **`salinita_dellorizzonte_superficiale_mscm__125`** | Polygon | Topsoil electrical conductivity (0–30 cm depth), identifying soil salinity hazards near the surface. |
| **`salinita_dellorizzonte_sottosuperficiale_1m_mscm__125`** | Polygon | Deep subsoil electrical conductivity (30–100 cm depth), detailing salinity risks down to a 1-meter depth profile. |
| **`erosione_potenziale_tha`** | Polygon | Modeled baseline potential soil loss from rainfall and runoff, quantified in metric tons per hectare per year. |
| **`capacita_duso_e_fertilita_dei_suoli`** | Polygon | Land Capability Classification ($LCC$) grading land from Class 1 (no agricultural limitations) to Class 8 (severe environmental limits). |

### **9. Socio-Economic Indicators, Vulnerability, & Municipal Metabolism**

*Municipal-level tax distribution, economic stratification, resource consumption, and structural vulnerability indices. CF is the number of contributors CR is the monetary value*

| Layer Name | Geometry | Conceptual Description |
| --- | --- | --- |
| **`tabular_comunal_census`** | Polygon | **Historical Resident Population Baseline:** Longitudinal demographic tracking capturing the total registered resident population as of December 31st (`RESPOP_AV`). Serves as the critical baseline denominator for calculating local per-capita exposure metrics and demographic shifts. |
| **`tabular_comunal_economical_princ`** | Polygon | **Taxpayers and Main Income Categories:** Comprehensive tax profile accounting for total taxpayers (`NTAXP`), taxable bases (`TAXABINC`), and specific wealth source distributions including buildings (`CADINCR` and `CADINCF`), employment (`SUBEMPTRINCR` and `SUBEMPTRINCF`), pensions (`PENSINCR` and `PENSINCF`), and corporate structures (`ENTROAINCR`, `ENTROAINCF`). |
| **`tabular_comunal_economical_reddito`** | Polygon | **Income Stratification Brackets:** Population density breakdown divided into progressive financial tiers spanning from negative/zero earnings (`E_UN0`) up through low-to-mid ranges (`E0-10000`, `E10000-15000`, `E15000-26000`) up to high-earning classes (`E55000-75000`). |
| **`tabular_comunal_economical_distrib`** | Polygon | **Municipal Water Metabolism:** Utility network metrics tracking municipal water balance by crossing total volume injected into the network (`ACQ_IMM`) against actual billable water delivered to citizens (`ACQ_EROG`) in thousands of cubic meters. |
| **`tabular_comunal_economical_indice_comp`** | Polygon | **Composite Vulnerability & Infrastructure Index:** Multi-dimensional deprivation and resilience framework anchored by a macro-vulnerability decile (`COMP_FRAG_INDEX_DECILE`), landslide hazard indices (`LANDSLIDE_HAZARD_ZONES`) and land consumption (`LAND_CONSUMPTION`),  alongside local development vectors covering labor market strength (`EMPL_RATE_20_64`), educational gaps (`POP_25_64_UNDER_DIP_LOW_SEC_EDU`), demographic dependency burdens (`POP_DEPEN_INDEX_ADJ`), essential services accessibility (`INDEX_ACCES_ESSENT_SERVICES`), and low-productivity industrial density (`PERSEMP_LU_LOW_PRO_INDSERV_VENTILE`). |
| **`pendolarismo_inflow_comunal`** | Polygon | **Commuter Inflow Aggregation:** Municipal-level summary tracking total daily incoming work-related commuters (`inflow_total`). Serves as a dynamic daytime population proxy to model daytime economic magnetism, infrastructure load, and mobility pressure. |

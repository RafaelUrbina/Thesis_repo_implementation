Adopt a **Star Schema** or **Relational View** approach, keeping your dimensions decoupled until the exact moment you feed them into an analytical script or AI model.

Here is the architectural blueprint to build H3 Resolution 10 data cube without wasting memory.

---

### **1. The Core Architecture: The "Snowflake" Structure**

Maintain **three separate tiers of data**, all linked by your two primary keys: `H3_Index` and `Timestamp`.

#### **Tier A: The Static Base (Space Only)**

This table contains data that never changes, or changes so slowly (geology, soil) that it can be treated as permanent.

* **Primary Key:** `H3_Index`
* **Columns:** `Soil_Type_Toscana`, `Elevation_DTM`, `Slope`, `Is_Urban_Building_DSM`.
* **Storage:** Very small. You only have one row per hexagon.

#### **Tier B: The Coarse Temporal Base (Space + Year)**

This table captures macro-environmental trends that are updated annually.

* **Primary Keys:** `H3_Index` + `Year`
* **Columns:** `InSAR_Annual_Subsidence_Velocity`, `Land_Use_Class` (Corine Land Cover).
* **Storage:** Compact. (Number of Hexagons $\times$ 15 Years).

#### **Tier C: The Fine Temporal Base (Space + Month)**

This is your actual "Fact Table" where the high-frequency dynamics live.

* **Primary Keys:** `H3_Index` + `Year_Month`
* **Columns:** `Monthly_Rain_Total` (from ISPRA/SIR), `Hydrological_Stress_Index` (your custom formula), `Temperature_Anomaly`.
* **Storage:** This is your largest table, but because it *only* contains the changing weather metrics, it stays incredibly lightweight.

---

### **2. The Parsimonious Processing Pipeline**

When you write your Python logic, you do not merge these tables permanently on disk. You keep them separate and use **"Just-In-Time" Broadcast Joins** or simple dictionary mapping.

Here is how you handle each specific temporal cadence:

#### **How to handle "Unique values of just one year" (e.g., Soil Map)**

Treat it as a **Static Broadcast**. In your final machine learning or risk scoring pipeline, you simply map the static `H3_Index` to your dynamic rows. If an H3 cell is `8b1f1...`, it always gets assigned "Unstable Clay," whether the row is for January 2011 or December 2025.

#### **How to handle "Yearly" data (e.g., InSAR ground movement)**

Use a **Temporal Forward-Fill or Match**. When analyzing a specific month (e.g., *October 2018*), your script pulls the annual subsidence value for the year *2018*.

* *Why this is parsimonious:* You only store the number "2018" once per hexagon, rather than repeating the string or float 12 times for every month of that year.

#### **How to handle "Monthly" data (Your Core Clock)**

Your data cube's "heartbeat" should be set to the finest temporal resolution you actually care about—which you've determined is **Monthly**.

---

### **3. Final Logical Schema of your Cube**

To visualize how this looks when your model processes it, think of it as a logical view rather than a physical file:

| H3_Index (Res 10) | Timestamp (Year_Month) | Monthly Rain (Dynamic) | Annual Subsidence (Year-Matched) | Soil Type (Static-Mapped) |
| --- | --- | --- | --- | --- |
| `8b1f1...` | `2025_01` | 120mm | -4.2mm/yr *(from 2025)* | Argille Scagliose |
| `8b1f1...` | `2025_02` | 45mm | -4.2mm/yr *(from 2025)* | Argille Scagliose |
| `8b1f1...` | `2026_01` | 98mm | -5.1mm/yr *(from 2026)* | Argille Scagliose |

### **Summary of Rules for Parsimony**

1. **Never duplicate strings:** Keep soil classes, municipality names, and geological units in a separate static lookup table. Only store an integer ID or the H3 index.
2. **Use Parquet or HDF5 format:** Do not use CSVs for this. Storing this data in Apache Parquet format allows you to compress columns aggressively (especially static ones that repeat), saving gigabytes of disk space.
3. **Aggregate before joining:** Ensure your rain data is already compressed to its monthly totals *before* you introduce it to the H3 grid system.

By keeping the spatial base separate from the temporal updates, you can run complex risk calculations across 15 years of history on a standard laptop without running out of RAM.

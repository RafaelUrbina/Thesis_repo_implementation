# Thesis Proposal: A Multimodal Data-Driven Framework for Water Distribution Network Resilience
**Case Study:** Province of Tuscany, Italy

## 1. Research Innovation: The "Tri-Layer" Methodology
This research proposes a transition from traditional static pipeline mapping to a dynamic, risk-aware framework by integrating three distinct knowledge layers:

1.  **The Detection Layer (Open Data & Remote Sensing):** Utilizing Sentinel-2 multitemporal imagery and regional Digital terrain models (DTMs) to identify "soil marks" and "vegetation indices" that reveal buried infrastructure and potential moisture anomalies.
2.  **The Vulnerability Layer (Environmental Risk):** Cross-referencing detected locations with hydrogeological risk maps (PAI) and subsidence data from the Tuscany Open Data portal.
3.  **The Mitigation Layer (Patent Intelligence):** Analyzing patent-protected technologies to identify the most effective hardware and logic for addressing specific detected risks.

---

## 2. Alignment with Research Questions & Objectives

### RQ1: Leveraging Open Data for Inefficiency Analysis
* **Application:** The study utilizes the **Geoportale Regionale della Toscana** and ESA’s **Sentinel Hub**.
* **Link:** By analyzing temporal variations in soil reflectance (Soil Marks) over 10 years (example), we identify pipeline segments where high non-revenue water (NRW) or physical leaks are statistically likely, turning raw Open Data into actionable infrastructure health indicators.

### RQ2: Multimodal Data Integration
* **Application:** Fusing **structured spatial data** (GIS layers of flood risks) (example) with **unstructured/semi-structured data** (satellite pixel values and technical patent descriptions) Or economic data (example).
* **Link:** The integration of L-Band SAR (microwave) with RGB/NIR imagery (examples) enhances detection accuracy in the complex clay-heavy soils, reducing "false positives" caused by surface agriculture. Mixing various environmental data can help in integrating a multidimensional risk score.

### RQ3: Utilizing Patent Solutions for Inefficiencies
* **Application:** Identifying patents for **self-healing pipes**, **non-invasive pressure sensors**, and **leak-detection algorithms**.
* **Link:** Patents are treated as "technical benchmarks." Instead of suggesting generic repairs, the framework recommends specific patented solutions (e.g., modular manifolds or acoustic sensors) (examples) based on the specific environmental risk identified (e.g., soil liquefaction in the province).

---

## 4. Expected Outcome: The Prioritization solution
The final output of the thesis is a **Decision Support Framework** for water utilities (e.g., Acque SpA). This framework will prioritize interventions by scoring pipeline segments based on:
* **Detection Confidence:** (How clearly the remote sensing shows a pipeline/leak).
* **Risk Severity:** (Probability of soil shift or flood damage in the province).
* **Technological Readiness:** (Availability of patented solutions to address that specific failure mode).

## 5. Summary for Professors
The core "concrete" value of this thesis lies in its **scalability**. By demonstrating that Open Data can identify buried assets and that Patent Data can define the solution set, we provide a low-cost, high-impact methodology for water managers to optimize resource allocation in the face of climate-driven environmental risks.

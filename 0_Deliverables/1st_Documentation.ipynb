    * [ISTAT](../../Thesis_repo/2_Data_extraction/0_preliminary_data_sources/ISTAT.ipynb) has a variety of metrics and time series related to regions, capitals or aggregated data that can be used, my only concern is the **veracity** of the information and the **velocity** since, first the data comes from surveys and tends to have NA's in some fields very frequently, second the data tends to be yearly or bianual with a small window in terms of time, usually about 10 years, meaninhg a window of about 10 to 20 time stamps. Disregarding the disadvantages in the link can be found the sources I found most useful and which have a potential to be integrated as "contextual information" in the model.

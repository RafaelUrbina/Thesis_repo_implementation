# Methodology proposal

## Model the provincial map 

### Structural set up

- Raster or Hexagonal grids (H3 or H2)
- Spatial joins would assign values to these regions in order to have attributes associated to locations from various sources

### Value assignments

- Continuous data (Aggregation methods to produce the most useful concrete forms of data)(Distance to events, Averages, Total counts in the vicinity, total events in the adjacent places...so on)
- Categorical data (soils, comune or provincia, has access to lake or sea, has access to diverse other risk factors...so on)

### Useful data

- Landslides
- Elevation, Digital terrain model (STANDARD DEVIATION OF ELEVATION PER HEXAGON, AVERAGE HEIGHT, STEEPEST ANGLE IN THE HEXAGON)
- Seismic historical data
- Historical data for substantial rain events
- Floods data
- Soil type datasets
- Economical data (Istat comunal level, discouraged prov level) - (Contribuenti e principali categorie di reddito, Contribuenti e reddito complessivo per classi di importo, Indice composito di fragilità e sue componenti - livello comunale, Distribuzione di acqua potabile)
- Social data (Istat demographics, demo.istat.it, basi territoriali cartografia, Popolazione residente ricostruita - Anni 2002-2019, Previsioni della popolazione  - Anni 2018-2065, )
- Water used
- Sewage used
- Hydrography data (check better)
- Street maps
- Diverse infrastructure classified (check better)

### Examples

- Seismic Stress	Distance to nearest historical epicenter / Magnitude
- Hydraulic Stress	Days of rain > 50mm in last 10 years
- Geological Risk	Distance to the edge of a Landslide Zone
- Course Proximity	Vertical distance to the nearest water course

### Tools to use while creating the data cube and training

- Spatial Lag features
- Spatial cross validation

### Weakly Supervised (Heuristic Flags)

- Defining risk as a formula of F(Slope, Seismic_events_near, Number_floods_near, Population_near, Roads_near, Soil_type, landslide_history......)
- Do spatial validations: similar train and test "risk" areas

### Cross check

- with historical data of damages in the water system, check if those areas are flagged as high risk areas for the model

### Bridging the distance between risk and solutions

- Use XAI models to explain the BB model solutions (SHAP, LIME,...)
- Pass this XAI model solutions as a targeted patent search query inside the patent pool
- This avoids mismatched solutions, justifies each component of the risk, identifies the interaction effects between features

### Satellite imagery

- Sentinel-1 VV Backscatter(surface water or excavation scar), Dual-Pol Ratio (Proxy soil moisture (dielectric signal)), Radar vegetation index
- Sentinel-2 NDVI, SAVI(soil adjusted) vegeation index, NMDI unusual moisture detection, NDWI surface moisture
- Challenge: Fuse satellite model results into the risk analysis
